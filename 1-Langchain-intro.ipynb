## streaming and batch


- streaming: output shown in chunks

```python
for chunks in model.stream("write 500 word essay on Ai"):
    print(chunks.text,end="",flush=True)
```

- batch: collection of independent requests

```python
response=model.batch([
    "why humans cant fly?",
    "why parrots have feathers",
    "How an aeroplane can fly"

    config={
        max_concurrency: 5, # max limit is 5
    }
])

for response in responses:
    print(respone)
```
