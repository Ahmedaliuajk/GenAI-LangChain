## Agents with Memory

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
```

```python
agent=create_agent(

    model="gpt-5",
    checkpointer=InMemorySaver()
)
```

```python
# run with Thread Id

config={"configurable":{"thread_id":"test1"}}
```

```python
from langchain.core_messages import HumanMessage,SystemMessage

response=agent.invoke({"messages":[HumanMessage(content="Hi I am Ahmed")]},config=config)
```
