## Tools


```python
import os
from langchain.chat_models import init_chat_model
os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
model=init_chat_model("gpt-5")
response=model.invoke("Write an essay on Ai")
response
```

## Tool creation

```python
from langchain.tool import tool

@tool

def get_weather(location:str)->str:
    """ What is the weather at location"""
    return f"the weather is sunny in{location}"


# now we will connect llm to tool

model_with_tools=model.bind_tool([get_weather])
```

```python
respone=model_with_tools.invoke("what is weather in Muzaffarabad")
respone
```

```python
respone.tool_calls
```

# Tool Execution Loop

```python
# Step-1: model generates tool call
messages=({"role":"user","content": "whats the weather in Muzaffarabad"})
ai_message = model_with_tools.invoke(messages)
messages.append(ai_message)

# Step-2:Execute tool and collect result
for tool_call in ai_message.tool_calls:
    tool_result=get_weather.invoke(tool_call)
    messages.append(tool_call)


# Step-3: pass result back to model for final response

final_response=model_with_tools.invoke(messages)
print(messages.text)
```
