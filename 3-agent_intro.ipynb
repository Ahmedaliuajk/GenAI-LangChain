## Creating Agent

```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["OPENAI_API_KEY"]=os.getenv("OPENAI_API_KEY")
```

```python
from langchain.tool import tool
@tool
def get_weather(location:str)->str:
    """ What is the weather at location"""
    return f"the weather is sunny in{location}"
```

```python
from langchain.agent import create_agent

agent=create_agent(
    model="gpt-5",
    tools=[getweather],
    system_prompt="you are an assistant"
)

agent
```

```python
respone=agent.invoke({"messages":[{"role":"user","content":"Whats the weather in mzd"}]})
```
